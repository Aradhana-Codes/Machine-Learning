```python
import pandas as pd
import requests
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
url="https://api.themoviedb.org/3/movie/top_rated"

```

```python
params={"api_key": os.getenv("TMDB_API_KEY")}
```

```python
response=requests.get(url, params=params)
```

```python
data=response.json()
```

```python
data
```

```python
df=pd.DataFrame(data['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
```

```python
df.head()
```

```python
for i in range(1, 491):
    url="https://api.themoviedb.org/3/movie/top_rated"

    params={"api_key": os.getenv("TMDB_API_KEY")}
    
    response= requests.get(url, params=params)
    data=response.json()
    temp_df=pd.DataFrame(data['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
    df = pd.concat([df, temp_df])  
```
